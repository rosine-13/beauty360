```python
import os

DATA_DIR = r"C:\beauty\beauty360\data\images\Oily-Dry-Skin-Types"
IMG_SIZE = (224, 224)   # taille attendue par MobileNetV2
BATCH_SIZE = 32         # nombre d'images par paquet
```

```python
train_ds = tf.keras.utils.image_dataset_from_directory(
    os.path.join(DATA_DIR, "train"),
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical",
    shuffle=True,
    seed=42
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    os.path.join(DATA_DIR, "valid"),
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical",
    shuffle=False
)

test_ds = tf.keras.utils.image_dataset_from_directory(
    os.path.join(DATA_DIR, "test"),
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical",
    shuffle=False
)

class_names = train_ds.class_names
print("Classes :", class_names)
```

```python
# ---- 1. AUGMENTATION : créer des variations des photos (seulement à l'entraînement)
augmentation = tf.keras.Sequential([
    tf.keras.layers.RandomFlip("horizontal"),   # effet miroir
    tf.keras.layers.RandomRotation(0.1),        # petite rotation
    tf.keras.layers.RandomZoom(0.1),            # petit zoom
    tf.keras.layers.RandomBrightness(0.1),      # plus clair / plus sombre
], name="augmentation")

# ---- 2. POIDS DES CLASSES : les classes rares comptent plus
counts = {c: len(os.listdir(os.path.join(DATA_DIR, "train", c))) for c in class_names}
total = sum(counts.values())
class_weight = {i: total / (len(class_names) * counts[c])
                for i, c in enumerate(class_names)}

print("Images par classe :", counts)
print("Poids :", class_weight)

# ---- 3. PREFETCH : préparer le paquet suivant à l'avance (plus rapide)
AUTOTUNE = tf.data.AUTOTUNE
train_ds = train_ds.prefetch(AUTOTUNE)
val_ds = val_ds.prefetch(AUTOTUNE)
test_ds = test_ds.prefetch(AUTOTUNE)
```

```python
# 1. Le modèle pré-entraîné, sans sa couche de décision d'origine
base = tf.keras.applications.MobileNetV2(
    input_shape=IMG_SIZE + (3,),   # images 224×224 en couleur (3 = rouge, vert, bleu)
    include_top=False,             # on enlève sa décision d'origine (1 000 objets)
    weights="imagenet")            # on garde ce qu'il a appris sur 1,4 million d'images
base.trainable = False             # on gèle ses connaissances

# 2. Notre modèle, étape par étape (comme le schéma)
inputs = tf.keras.Input(shape=IMG_SIZE + (3,))
x = augmentation(inputs)
x = tf.keras.applications.mobilenet_v2.preprocess_input(x)
x = base(x, training=False)
x = tf.keras.layers.GlobalAveragePooling2D()(x)
x = tf.keras.layers.Dropout(0.3)(x)
outputs = tf.keras.layers.Dense(3, activation="softmax")(x)
model = tf.keras.Model(inputs, outputs)

# 3. Les règles d'apprentissage
model.compile(
    optimizer=tf.keras.optimizers.Adam(1e-3),  # la méthode pour corriger ses erreurs
    loss="categorical_crossentropy",            # la façon de mesurer ses erreurs
    metrics=["accuracy"])                       # ce qu'on affiche : % de bonnes réponses

model.summary()
```

```python
# Arrêt automatique + on garde la meilleure version du modèle
stop = tf.keras.callbacks.EarlyStopping(
    monitor="val_accuracy",        # on surveille le score sur la validation
    patience=3,                    # on tolère 3 époques sans progrès
    restore_best_weights=True)     # à la fin, on reprend la meilleure version

history = model.fit(
    train_ds,                      # les photos pour apprendre
    validation_data=val_ds,        # les photos pour surveiller
    epochs=15,                     # 15 passages maximum
    class_weight=class_weight,     # plus d'importance aux peaux sèches
    callbacks=[stop])
```

```python
os.makedirs(r"C:\beauty\beauty360\models", exist_ok=True)   # crée le dossier models

# 1. Dégeler les dernières couches de MobileNetV2
base.trainable = True
for layer in base.layers[:-40]:   # toutes les couches SAUF les 40 dernières...
    layer.trainable = False       # ...restent gelées

# 2. Recompiler avec un apprentissage très doux
model.compile(
    optimizer=tf.keras.optimizers.Adam(1e-5),   # 100 fois plus doux qu'avant
    loss="categorical_crossentropy",
    metrics=["accuracy"])

# 3. Sauvegarde automatique du meilleur modèle à chaque progrès
sauvegarde = tf.keras.callbacks.ModelCheckpoint(
    r"C:\beauty\beauty360\models\skin_model.keras",
    monitor="val_accuracy",
    save_best_only=True)

stop = tf.keras.callbacks.EarlyStopping(
    monitor="val_accuracy", patience=4, restore_best_weights=True)

# 4. Entraîner à nouveau
history_fine = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=15,
    class_weight=class_weight,
    callbacks=[stop, sauvegarde])
```
